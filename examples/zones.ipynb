### SFOS API - Network Zones Management

**Purpose:** Create, read, update, and delete network zones

**Topics:**
- Basic CRUD operations using domain StrEnum objects
- Pagination and filtering
- Error handling
- Bulk operations

**Requirements:**
- Environment variables configured (.env file)
- SFOS firewall accessible
- API key with appropriate permissions

---

#### Setup: Import Libraries and Configure Client

```python
from common import print_data, setup_client

from sfos_api import Network

# Initialize the SFOS API Client
client = setup_client()
print("✅ SFOS API client initialized")
```

#### Create Network Zones

Create network zone objects using the `Network.ZONES` endpoint object.

Zones are used to group network interfaces and define traffic policies.

API Link: [Create a zone | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/create-zone/)

```python
# Create a network zone
endpoint = Network.ZONES
zone_name = "test_zone_dmz"

payload = {
    "name": zone_name,
    "description": "DMZ zone for hosting services",
    "type": "dmz",
}

try:
    response = client.create(endpoint, payload)
    print(f"✅ Created zone '{zone_name}'")
    print_data(response)
except Exception as e:
    print(f"Error creating zone: {e}")
    print_data(str(e))
```

```python
# Create another network zone
endpoint = Network.ZONES
zone_name = "test_zone_guest"

payload = {
    "name": zone_name,
    "description": "Guest network zone",
    "type": "lan",
}

try:
    response = client.create(endpoint, payload)
    print(f"✅ Created zone '{zone_name}'")
    print_data(response)
except Exception as e:
    print(f"Error creating zone: {e}")
    print_data(str(e))
```

#### Read Network Zones

Retrieve zone information by ID/name or list all zones.

API Links: [Get all zones | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/get-all-zones/) | [Get a zone | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/get-zone/)

```python
# Get a specific zone by name
endpoint = Network.ZONES
zone_name = "test_zone_guest"

try:
    response = client.read(endpoint, id_or_name=zone_name)
    print(f"✅ Retrieved zone '{zone_name}'")
    print_data(response)
except Exception as e:
    print(f"Error reading zone: {e}")
    print_data(str(e))
```

```python
# Get all zones
endpoint = Network.ZONES

try:
    response = client.read(endpoint)

    if "items" in response and response["items"]:
        print(f"✅ Retrieved {len(response['items'])} zones")
        print_data(response["items"], keys=["name", "description", "type"])
    else:
        print("No zones found")
except Exception as e:
    print(f"Error reading zones: {e}")
    print_data(str(e))
```

```python
# Get zones with filtering
endpoint = Network.ZONES
params = {
    "nameContains": "test",
}

try:
    response = client.read(endpoint, params=params, pageTotal=False)

    if "items" in response:
        print(f"✅ Retrieved {len(response['items'])} zones containing 'test-'")
        print_data(response["items"], keys=["name", "description", "type"])
    else:
        print("No zones found matching filter")
except Exception as e:
    print(f"Error reading zones: {e}")
    print_data(str(e))
```

#### Update Network Zone

Modify an existing zone's properties.

API Link: [Update a zone | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/update-zone/)

```python
# Update a zone
endpoint = Network.ZONES
zone_name = "test_zone_dmz"

payload = {
    "description": "Updated DMZ zone - for web and app servers",
    "type": "lan",
    "services": [
        {"name": "ssh"},
        {"name": "https"},
    ],
}

try:
    response = client.update(endpoint, id_or_name=zone_name, data=payload)
    print(f"✅ Updated zone '{zone_name}'")
    print_data(response)
except Exception as e:
    print(f"Error updating zone: {e}")
    print_data(str(e))
```

#### Delete Network Zones

Remove zone objects (single or bulk delete).

API Links: [Delete a zone | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/delete-zone/) | [Bulk delete zones | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/bulk-delete-zones/)

```python
# Delete a single zone
endpoint = Network.ZONES
zone_name = "test_zone_dmz"

try:
    response = client.delete(endpoint, id_or_name=zone_name)
    print(f"✅ Deleted zone '{zone_name}'")
    if response:
        print_data(response)
except Exception as e:
    print(f"Error deleting zone: {e}")
    print_data(str(e))
```

```python
# Bulk delete zones
endpoint = Network.ZONES

items_to_delete = [
    {"name": "test_zone_dmz"},
    {"name": "test_zone_guest"},
]

try:
    response = client.delete(endpoint, items=items_to_delete)
    print(f"✅ Bulk deleted {len(items_to_delete)} zones")
    if response:
        print_data(response)
except Exception as e:
    print(f"Error bulk deleting zones: {e}")
    print_data(str(e))
```

```python
# Bulk delete zones with filter
endpoint = Network.ZONES

try:
    response = client.delete(endpoint, nameContains="test_")
    print("✅ Bulk deleted zones matching nameContains='test_'")
    if response:
        print_data(response)
except Exception as e:
    print(f"Error bulk deleting zones: {e}")
    print_data(str(e))
```

```python
# Close the client session when done
client.close()
print("✅ Client session closed")
```
