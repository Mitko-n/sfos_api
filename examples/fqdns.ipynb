### SFOS API - FQDN Address Objects Management

**Purpose:** Create, read, update, and delete FQDN (Fully Qualified Domain Name) address objects

**Topics:**
- Basic CRUD operations using domain StrEnum objects
- FQDN object creation
- Pagination and filtering
- Error handling
- Bulk operations

**Requirements:**
- Environment variables configured (.env file)
- SFOS firewall accessible
- API key with appropriate permissions

---

#### Setup: Import Libraries and Configure Client

```python
from common import print_data, setup_client

from sfos_api import Network

# Initialize the SFOS API Client
client = setup_client()
print("✅ SFOS API client initialized")
```

#### Create FQDN Address Objects

Create FQDN address objects using the `Network.ADDRESSES_FQDN` endpoint object.

FQDN objects allow you to reference hosts by domain name instead of IP addresses.

API Link: [Create a FQDN | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/create-fqdn/)

```python
# Create an FQDN address object
endpoint = Network.ADDRESSES_FQDN
fqdn_name = "test-fqdn-google"

payload = {
    "name": fqdn_name,
    "description": "Google DNS server",
    "fqdn": "dns.google.com",
}

try:
    response = client.create(endpoint, payload)
    print(f"✅ Created FQDN '{fqdn_name}'")
    print_data(response)
except Exception as e:
    print(f"Error creating FQDN: {e}")
    print_data(str(e))
```

```python
# Create another FQDN address object
endpoint = Network.ADDRESSES_FQDN
fqdn_name = "test-fqdn-cloudflare"

payload = {
    "name": fqdn_name,
    "description": "Cloudflare DNS server",
    "fqdn": "1dot1dot1dot1.cloudflare-dns.com",
}

try:
    response = client.create(endpoint, payload)
    print(f"✅ Created FQDN '{fqdn_name}'")
    print_data(response)
except Exception as e:
    print(f"Error creating FQDN: {e}")
    print_data(str(e))
```

```python
# Create an FQDN with a subdomain
endpoint = Network.ADDRESSES_FQDN
fqdn_name = "test-fqdn-company-api"

payload = {
    "name": fqdn_name,
    "description": "Company API server",
    "fqdn": "api.company.example.com",
}

try:
    response = client.create(endpoint, payload)
    print(f"✅ Created FQDN '{fqdn_name}'")
    print_data(response)
except Exception as e:
    print(f"Error creating FQDN: {e}")
    print_data(str(e))
```

#### Read FQDN Address Objects

Retrieve FQDN address object information by ID/name or list all FQDN objects.

API Links: [Get all FQDNs | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/get-all-fqdns/) | [Get a FQDN | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/get-fqdn-by-id-or-name/)

```python
# Get a specific FQDN address by name
endpoint = Network.ADDRESSES_FQDN
fqdn_name = "test-fqdn-google"

try:
    response = client.read(endpoint, id_or_name=fqdn_name)
    print(f"✅ Retrieved FQDN '{fqdn_name}'")
    print_data(response)
except Exception as e:
    print(f"Error reading FQDN: {e}")
    print_data(str(e))
```

```python
# Get all FQDN addresses
endpoint = Network.ADDRESSES_FQDN

try:
    response = client.read(endpoint)

    if "items" in response and response["items"]:
        print(f"✅ Retrieved {len(response['items'])} FQDN addresses")
        print_data(response["items"])
    else:
        print("No FQDN addresses found")
except Exception as e:
    print(f"Error reading FQDN addresses: {e}")
    print_data(str(e))
```

```python
# Get FQDN addresses with filtering
endpoint = Network.ADDRESSES_FQDN
params = {
    "nameContains": "test-fqdn-",
}

try:
    response = client.read(endpoint, params=params)

    if "items" in response:
        print(f"✅ Retrieved {len(response['items'])} FQDN addresses containing 'test-fqdn-'")
        print_data(response["items"], keys=["name", "description"])
    else:
        print("No FQDN addresses found matching filter")
except Exception as e:
    print(f"Error reading FQDN addresses: {e}")
    print_data(str(e))
```

#### Update FQDN Address Object

Modify an existing FQDN address object's properties.

API Link: [Update a FQDN | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/update-fqdn-by-id-or-name/)

```python
# Update an FQDN address
endpoint = Network.ADDRESSES_FQDN
fqdn_name = "test-fqdn-google"

payload = {
    "description": "Google public DNS server (updated)",
    "fqdn": "8.8.8.8.dns.google.com",
}

try:
    response = client.update(endpoint, id_or_name=fqdn_name, data=payload)
    print(f"✅ Updated FQDN '{fqdn_name}'")
    print_data(response)
except Exception as e:
    print(f"Error updating FQDN: {e}")
    print_data(str(e))
```

#### Delete FQDN Address Objects

Remove FQDN address objects (single or bulk delete).

API Links: [Delete a FQDN | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/delete-fqdn-by-id-or-name/) | [Bulk delete FQDNs | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/bulk-delete-fqdns/)

```python
# Delete a single FQDN address
endpoint = Network.ADDRESSES_FQDN
fqdn_name = "test-fqdn-google"

try:
    response = client.delete(endpoint, id_or_name=fqdn_name)
    print(f"✅ Deleted FQDN '{fqdn_name}'")
    if response:
        print_data(response)
except Exception as e:
    print(f"Error deleting FQDN: {e}")
    print_data(str(e))
```

```python
# Bulk delete FQDN addresses
endpoint = Network.ADDRESSES_FQDN

items_to_delete = [
    {"name": "test-fqdn-google"},
    {"name": "test-fqdn-cloudflare"},
    {"name": "test-fqdn-company-api"},
]

try:
    response = client.delete(endpoint, items=items_to_delete)
    print(f"✅ Bulk deleted {len(items_to_delete)} FQDN addresses")
    if response:
        print_data(response)
except Exception as e:
    print(f"Error bulk deleting FQDN addresses: {e}")
    print_data(str(e))
```

```python
# Close the client session when done
client.close()
print("✅ Client session closed")
```
