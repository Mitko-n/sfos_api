### SFOS API Client - IPv4 Address CRUD Operations

✅ **This notebook demonstrates how to use the SFOS API with the `SFOSClient` class using the new `Endpoints` notation.**

All CRUD operations are simplified through the client's automatic pagination and error handling, providing a clean Pythonic interface to the Sophos Firewall REST API.

```python
from common import print_data, setup_client

from sfos_api import Network

# Initialize the SFOS API Client
client = setup_client()
```

#### Create IPv4 Address

Create various types of IPv4 address objects (single IP, network, range, or list) using the `Network.ADDRESSES_IPV4` endpoint object.

API Link: [Create an IPv4 address | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/create-ip-address/)


```python
# Create IPv4 Address using Endpoint object
endpoint = Network.ADDRESSES_IPV4
ipv4_address = "test-ipv4address-one"

payload = {
    "name": ipv4_address,
    "description": "This is created by SFOS API",
    "type": "ipv4Address",
    "ipv4Address": "198.51.100.1",
}

try:
    response = client.create(endpoint, payload)
    print_data(response)
except Exception as e:
    print(f"Error creating IPv4 address: {e}")
    print_data(str(e))
```

#### Get All IPv4 Addresses

Retrieve all IPv4 address objects from the firewall with automatic pagination.

API Links: [Get all IPv4 addresses | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/get-all-ip-addresses/) | [Get an IPv4 address | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/get-ip-address-by-id-or-name/)

```python
# Get all IPv4 addresses
endpoint = Network.ADDRESSES_IPV4

try:
    response = client.read(endpoint)

    if "items" in response and response["items"]:
        print(f"✅ Retrieved {len(response['items'])} IPv4 addresses")
        print_data(response["items"])
    else:
        print("No IPv4 addresses found")
except Exception as e:
    print(f"Error reading IPv4 addresses: {e}")
    print_data(str(e))
```

```python
# Get IPv4 addresses by name filter (using params dict)
endpoint = Network.ADDRESSES_IPV4
params = {
    "nameContains": "test-",
    "type": "ipv4Network",
}

try:
    response = client.read(endpoint, params=params)

    if "items" in response and response["items"]:
        print(f"✅ Retrieved {len(response['items'])} addresses containing 'test-'")
        print_data(response["items"], keys=["name", "description", "type"])
    else:
        print("No addresses found matching filter")
except Exception as e:
    print(f"Error reading IPv4 addresses: {e}")
    print_data(str(e))
```

```python
# Get IPv4 addresses with pagination (using params dict)
endpoint = Network.ADDRESSES_IPV4
params = {
    "page": 1,
    "pageSize": 12,
}

try:
    response = client.read(endpoint, params=params)

    if "items" in response:
        pages_info = response.get("pages", {})
        current_page = pages_info.get("current", response.get("page", 1))
        total_pages = pages_info.get("total", 1)
        print(f"✅ Page {current_page} of {total_pages}")
        print(f"   Total items on this page: {len(response['items'])}")
        print_data(response["items"][:5], keys=["name", "description", "type"])
        if len(response["items"]) > 5:
            print(f"   ... (showing first 5 of {len(response['items'])} items)")
    else:
        print_data(response)
except Exception as e:
    print(f"Error reading IPv4 addresses: {e}")
    print_data(str(e))
```

```python
# Request only a specific page (no auto-pagination)
endpoint = Network.ADDRESSES_IPV4

try:
    response = client.read(endpoint, page=2, pageSize=5)

    if "items" in response:
        pages_info = response.get("pages", {})
        current_page = pages_info.get("current", response.get("page", 1))
        items_count = len(response["items"])
        print(f"✅ Page {current_page} only: {items_count} items on this page")
        print_data(response["items"])
    else:
        print_data(response)
except Exception as e:
    print(f"Error reading IPv4 addresses: {e}")
    print_data(str(e))
```

```python
# Keyword arguments take precedence over params dict
endpoint = Network.ADDRESSES_IPV4
params = {
    "page": 1,
    "pageSize": 50,
    "nameContains": "test-",
}

try:
    response = client.read(endpoint, params=params, pageSize=5)

    if "items" in response:
        pages_info = response.get("pages", {})
        current_page = pages_info.get("current", response.get("page", 1))
        print(f"✅ Retrieved page {current_page} only (not auto-paginated)")
        print_data(response["items"])
    else:
        print_data(response)
except Exception as e:
    print(f"Error reading IPv4 addresses: {e}")
    print_data(str(e))
```

```python
# Using keyword arguments (recommended approach)
# Auto-paginate and retrieve all items
endpoint = Network.ADDRESSES_IPV4

try:
    response = client.read(endpoint, page=1, pageSize=12)

    if "items" in response:
        print(f"✅ Retrieved all pages: {len(response['items'])} items total")
        print_data(response["items"], keys=["name", "description", "type"])
        if len(response["items"]) > 5:
            print(f"... (showing first 5 of {len(response['items'])} items)")
    else:
        print_data(response)
except Exception as e:
    print(f"Error reading IPv4 addresses: {e}")
    print_data(str(e))
```

#### Update IPv4 Address

Modify an existing IPv4 address object's properties.

API Link: [Update an IPv4 address | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/update-ip-address-by-id-or-name/)

```python
# Update IPv4 address
endpoint = Network.ADDRESSES_IPV4
id_or_name = "test-ipv4address-one"

payload = {
    "description": "Updated by SFOS API",
    "type": "ipv4Address",
    "ipv4Address": "198.51.100.50",
}

try:
    response = client.update(endpoint, id_or_name=id_or_name, data=payload)
    print(f"✅ Updated address '{id_or_name}'")
    print_data(response)
except Exception as e:
    print(f"Error updating IPv4 address: {e}")
    print_data(str(e))
```

#### Delete IPv4 Addresses

Remove IPv4 address objects (single or bulk delete).

API Links: [Delete an IPv4 address | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/delete-ip-address-by-id-or-name/) | [Bulk delete IPv4 | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/bulk-delete-ip-addresses/)

```python
# Delete a single IPv4 address
endpoint = Network.ADDRESSES_IPV4
id_or_name = "test-ipv4address-one"

try:
    response = client.delete(endpoint, id_or_name=id_or_name)
    print(f"✅ Deleted address '{id_or_name}'")
    if response:
        print_data(response)
except Exception as e:
    print(f"Error deleting IPv4 address: {e}")
    print_data(str(e))
```

#### Bulk delete IPv4 addresses

Delete multiple IPv4 address objects in a single operation.

API Link: [Bulk delete IPv4 | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/bulk-delete-ip-addresses/)

```python
# Bulk delete IPv4 addresses
endpoint = Network.ADDRESSES_IPV4

items_to_delete = [
    {"name": "test-ipv4address-one"},
    {"name": "test-ipv4address-two"},
    {"name": "test-ipv4network"},
    {"name": "test-ipv4range"},
    {"name": "test-ipv4list"},
]

try:
    response = client.delete(endpoint, items=items_to_delete)
    print(f"✅ Bulk deleted {len(items_to_delete)} addresses")
    if response:
        print_data(response)
except Exception as e:
    print(f"Error bulk deleting IPv4 addresses: {e}")
    print_data(str(e))
```
