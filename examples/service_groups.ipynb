### SFOS API - Service Groups Management

**Purpose:** Create, read, update, and delete service groups

**Topics:**
- Basic CRUD operations using domain StrEnum objects
- Group membership management
- Pagination and filtering
- Error handling
- Bulk operations

**Requirements:**
- Environment variables configured (.env file)
- SFOS firewall accessible
- API key with appropriate permissions

---

#### Setup: Import Libraries and Configure Client

```python
from common import print_data, setup_client

from sfos_api import Network

# Initialize the SFOS API Client
client = setup_client()
print("✅ SFOS API client initialized")
```

#### Create Service Groups

Create service group objects using the `Network.SERVICE_GROUPS` endpoint object.

Service groups allow you to combine multiple services for use in firewall rules.

API Link: [Create a Service group | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/create-service-group/)

---

```python
# Create a service group
endpoint = Network.SERVICE_GROUPS
group_name = "test-service-group-web"

payload = {
    "name": group_name,
    "description": "Web services group",
    "services": [
        {"name": "HTTP"},
        {"name": "HTTPS"},
    ],
}

try:
    response = client.create(endpoint, payload)
    print(f"✅ Created service group '{group_name}'")
    print_data(response)
except Exception as e:
    print(f"Error creating service group: {e}")
    print_data(str(e))
```

```python
# Create another service group
endpoint = Network.SERVICE_GROUPS
group_name = "test-service-group-email"

payload = {
    "name": group_name,
    "description": "Email services group",
    "services": [
        {"name": "SMTP"},
        {"name": "POP3"},
        {"name": "IMAP"},
    ],
}

try:
    response = client.create(endpoint, payload)
    print(f"✅ Created service group '{group_name}'")
    print_data(response)
except Exception as e:
    print(f"Error creating service group: {e}")
    print_data(str(e))
```

#### Read Service Groups

Retrieve service group information by ID/name or list all service groups.

API Links: [Get all service groups | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/get-all-service-groups/) | [Get a service group | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/get-service-group-by-id-or-name/)
____

```python
# Get a specific service group by name
endpoint = Network.SERVICE_GROUPS
group_name = "test-service-group-web"

try:
    response = client.read(endpoint, id_or_name=group_name)
    print(f"✅ Retrieved service group '{group_name}'")
    print_data(response)
except Exception as e:
    print(f"Error reading service group: {e}")
    print_data(str(e))
```

```python
# Get all service groups
endpoint = Network.SERVICE_GROUPS

try:
    response = client.read(endpoint)

    if "items" in response and response["items"]:
        print(f"✅ Retrieved {len(response['items'])} service groups")
        # Show first 5 groups
        print_data(response["items"], keys=["name", "description", "services"])
        if len(response["items"]) > 5:
            print(f"... (showing first 5 of {len(response['items'])} groups)")
    else:
        print("No service groups found")
except Exception as e:
    print(f"Error reading service groups: {e}")
    print_data(str(e))
```

```python
# Get service groups with filtering
endpoint = Network.SERVICE_GROUPS
params = {
    "nameContains": "test-service-group-",
    "page": 1,
    "pageSize": 10,
}

try:
    response = client.read(endpoint, params=params, pageTotal=False)

    if "items" in response:
        print(f"✅ Retrieved {len(response['items'])} service groups containing 'test-service-group-'")
        print_data(response["items"], keys=["name", "description", "services"])
    else:
        print("No service groups found matching filter")
except Exception as e:
    print(f"Error reading service groups: {e}")
    print_data(str(e))
```

#### Update Service Group

Modify an existing service group's properties.

API Link: [Update a service group | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/update-service-group-by-id-or-name/)
___

```python
# Update a service group
endpoint = Network.SERVICE_GROUPS
group_name = "test-service-group-web"

payload = {
    "description": "Web services group - HTTP, HTTPS, and SSH",
    "services": [
        {"name": "HTTP"},
        {"name": "HTTPS"},
        {"name": "SSH"},
    ],
}

try:
    response = client.update(endpoint, id_or_name=group_name, data=payload)
    print(f"✅ Updated service group '{group_name}'")
    print_data(response)
except Exception as e:
    print(f"Error updating service group: {e}")
    print_data(str(e))
```

#### Delete Service Groups

Remove service group objects (single or bulk delete).

API Links: [Delete a service group | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/delete-service-group-by-id-or-name/) | [Bulk delete groups | Sophos Firewall API](https://docs.sophos.com/nsg/sophos-firewall/rest-api/api/firewall-config-openapi/bulk-delete-service-groups/)
___

```python
# Delete a single service group
endpoint = Network.SERVICE_GROUPS
group_name = "test-service-group-web"

try:
    response = client.delete(endpoint, id_or_name=group_name)
    print(f"✅ Deleted service group '{group_name}'")
    if response:
        print_data(response)
except Exception as e:
    print(f"Error deleting service group: {e}")
    print_data(str(e))
```

```python
# Bulk delete service groups
endpoint = Network.SERVICE_GROUPS

items_to_delete = [
    {"name": "test-service-group-web"},
    {"name": "test-service-group-email"},
]

try:
    response = client.delete(endpoint, items=items_to_delete)
    print(f"✅ Bulk deleted {len(items_to_delete)} service groups")
    if response:
        print_data(response)
except Exception as e:
    print(f"Error bulk deleting service groups: {e}")
    print_data(str(e))
```

```python
# Close the client session when done
client.close()
print("✅ Client session closed")
```
